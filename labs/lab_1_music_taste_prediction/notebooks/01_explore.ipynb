# 01 Exploration

Thin on purpose: the functions live in `songtaste.explore`, this is the place to look at what they draw. `make data` first. The written reading is `results/01-exploration.md`.

```python
from songtaste.data import SPEC, load_training, load_test
from songtaste import explore

train, test = load_training(), load_test()
SPEC
```

## Class balance

```python
explore.plot_class_balance(train, SPEC)
explore.class_balance_table(train, SPEC)
```

## Numeric features by label

Each density normalised within its class, so the imbalance does not read as separation.

```python
explore.plot_numeric_by_label(train, SPEC)
explore.separation(train, SPEC)
```

## Categoricals as proportions by label

```python
explore.plot_categorical_by_label(train, SPEC)
explore.categorical_table(train, SPEC)
```

## Correlation

```python
explore.plot_correlation(train, SPEC);
```

## Oddities

```python
explore.outliers(train, SPEC)
```

```python
explore.duplicates(train, test, SPEC)
```

## Save every figure to `figures/`

```python
explore.save_all(train, SPEC)
```
