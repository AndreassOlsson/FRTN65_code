# 03 Compare

Thin on purpose: the sweep is `songtaste.report`, the methods `songtaste.models.REGISTRY`, the rule `results/protocol.md`. `make data` first. The written reading is `results/03-sweep.md`; `make sweep` writes the CSV this reads (about 40 minutes, so this notebook reads the saved run rather than redoing it).

```python
import pandas as pd
from songtaste import evaluate, report

rows = pd.read_csv(evaluate.RESULTS / '03-sweep.csv')
rows.groupby('method').size()
```

## The comparison

Mean and corrected standard error over the same 25 splits, best accuracy first.

```python
report.comparison(rows).round(3)
```

```python
report.plot_splits(rows);
```

## Paired against the best

The corrected resampled t-test and its Bayesian reading, as in sklearn's model-comparison example; `p_rope` is the posterior probability that the gap is inside one point of accuracy.

```python
report.paired(rows).round(4)
```

## The decision

The methods inside the band, ranked the way the rule ranks them, and the one it returns.

```python
choice, band = report.decide(rows)
choice, band.round(4)
```

```python
evaluate.chosen_params(rows).query('method == @choice')
```
